```python
import numpy as np
import sympy


def to_polar_deg(z, ndigits=3):
    r = float(sympy.N(sympy.Abs(z)))
    theta_deg = float(sympy.N(sympy.deg(sympy.arg(z))))
    return f"{r:.{ndigits}f} ∠ {theta_deg:.{ndigits}f}°"


def to_rect(z, ndigits=3):
    x = float(sympy.N(sympy.re(z)))
    y = float(sympy.N(sympy.im(z)))
    sign = "+" if y >= 0 else "-"
    return f"{x:.{ndigits}f} {sign} j{abs(y):.{ndigits}f}"
```

## problem 1

```python
V, Vs, ZL, ZC, omega = sympy.symbols("V Vs Z_L Z_C omega")

omega = 4
L = 1
C = 0.25
theta = np.radians(45)
Vs = 24 * np.exp(1j * theta)
ZL = 1j * omega * L
ZC = 1 / (1j * C * omega)

eq = sympy.Eq(3 / 2 * (Vs - V) / ZL, V / (1 + ZC))
V = sympy.solve(eq, V)[0]
print("V=", to_rect(V))
print("V=", to_polar_deg(V))
print()
vo = V * 1 / (1 + ZC)
print("vo=", to_rect(vo))
print("vo=", to_polar_deg(vo))
```

## problem 2

```python
def parallel(Z1, Z2):
    return Z1 * Z2 / (Z1 + Z2)


Vs, Z1, Z2, omega = sympy.symbols("V Z_1 Z_2 omega")

f = 5000
L1 = 0.001
L2 = 0.002
Vs = 10

omega = 2 * sympy.pi * f
Z1 = sympy.I * omega * L1
Z2 = sympy.I * omega * L2

Z3 = parallel(Z1, 10)
V3 = Vs * Z3 / Z1
vo = V3 * 50 / (50 + Z3 + Z2)

print("Vo = ", to_rect(vo.evalf()))
print("Vo = ", to_polar_deg(vo.evalf()))
```

```python
V0, V1 = sympy.symbols("V_0 V_1")
eq1 = sympy.Eq((10 - V1) / (10j * sympy.pi), V1 / 10 + (V1 - V0) / (20j * sympy.pi))
eq2 = sympy.Eq((V1 - V0) / (20j * sympy.pi), V0 / 50)
ans = sympy.solve((eq1, eq2), (V0, V1))
print("V_0 = ", to_rect(ans[V0]))
print("V_0 = ", to_polar_deg(ans[V0]))
display(eq2)
```

```python
Zth, Vth = sympy.symbols("Z_{th} V_{th}")

Zth = parallel(10, parallel(-20j, 10 + 10j))
print("Zth = ", to_rect(Zth))
print("Zth = ", to_polar_deg(Zth))
print()

eq = sympy.Eq(2, Vth / 10 + Vth / (-20j) + (Vth - 20) / (10 + 10j))
vo = sympy.solve(eq, Vth)[0]
print("Vth = ", to_rect(vo))
print("Vth = ", to_polar_deg(vo))
print()
print("IN = ", to_rect(vo / Zth))
print("IN = ", to_polar_deg(vo / Zth))
```
